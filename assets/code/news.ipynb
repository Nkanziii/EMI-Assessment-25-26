```python
from transformers import pipeline

emotion_model = pipeline(
    "text-classification",
    model="j-hartmann/emotion-english-distilroberta-base",
    top_k=None,
    device=-1
)
```

```python
for headline in headlines[:5]:
    result = emotion_model(headline)
    print(f"\n{headline}")
    print(result)
```
